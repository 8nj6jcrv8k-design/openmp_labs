```python
import pandas as pd
import matplotlib.pyplot as plt


# ГРАФИК УСКОРЕНИЯ И ГРАФИК ЭФФЕКТИВНОСТИ

data_eff = pd.read_csv('table2_efficiency.csv')
t1_time = 0.110952
df_spmd = data_eff[data_eff['Метод'] == 'SPMD'].sort_values('Потоки')
df_reduction = data_eff[data_eff['Метод'] == 'REDUCTION'].sort_values('Потоки')

threads = df_spmd['Потоки'].tolist()
time_spmd = df_spmd['Время'].tolist()
time_reduction = df_reduction['Время'].tolist()

speedup_spmd = [t1_time / t for t in time_spmd]
speedup_reduction = [t1_time / t for t in time_reduction]
ideal_speedup = threads

efficiency_spmd = [s / p for s, p in zip(speedup_spmd, threads)]
efficiency_reduction = [s / p for s, p in zip(speedup_reduction, threads)]
ideal_efficiency = [1.0] * len(threads)

plt.figure(figsize=(7, 4.5))
plt.plot(threads, ideal_speedup, color='gray', linestyle='--', label='Идеальное ускорение')
plt.plot(threads, speedup_spmd, color='blue', marker='o', linewidth=2, label='Метод SPMD')
plt.plot(threads, speedup_reduction, color='green', marker='s', linewidth=2, label='Метод REDUCTION')

plt.title('График ускорения программы ($S_p$)')
plt.xlabel('Число вычислительных потоков (p)')
plt.ylabel('Коэффициент ускорения ($S_p$)')
plt.xticks(threads)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig('lab2_graph_speedup.png', dpi=300)
plt.close()


plt.figure(figsize=(7, 4.5))
plt.plot(threads, ideal_efficiency, color='gray', linestyle='--', label='Идеальная эффективность')
plt.plot(threads, efficiency_spmd, color='blue', marker='o', linewidth=2, label='Метод SPMD')
plt.plot(threads, efficiency_reduction, color='green', marker='s', linewidth=2, label='Метод REDUCTION')

plt.title('График эффективности распараллеливания ($E_p$)')
plt.xlabel('Число вычислительных потоков (p)')
plt.ylabel('Коэффициент эффективности ($E_p$)')
plt.xticks(threads)
plt.ylim(0, 1.1) # Шкала от 0 до 110%
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.tight_layout()
plt.savefig('lab2_graph_efficiency.png', dpi=300)
plt.close()

# ГРАФИК ПЛАНИРОВЩИКА (Столбики)

data_sched = pd.read_csv('table3_schedule.csv')
options = [str(x) for x in data_sched['Parametr' if 'Parametr' in data_sched.columns else 'Параметр'].tolist()]
time_options = data_sched['Время'].tolist()

plt.figure(figsize=(6, 4.5))
plt.bar(options, time_options, color=['skyblue', 'dodgerblue', 'navy'], width=0.5)

plt.title('Влияние размера блока static на время работы (2 потока)')
plt.xlabel('Параметр планировщика (OMP_SCHEDULE)')
plt.ylabel('Время выполнения (сек)')
plt.grid(axis='y', linestyle=':', alpha=0.6)

min_time = min(time_options)
max_time = max(time_options)
plt.ylim(min_time * 0.98, max_time * 1.02) 

plt.tight_layout()
plt.savefig('lab2_graph_schedule.png', dpi=300)
plt.close()

```
